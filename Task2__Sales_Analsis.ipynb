```python
import numpy as np
import pandas as pd

# 1. Generate Cleaned Dataset
np.random.seed(42)
n_rows = 500
dates = pd.date_range("2024-01-01", periods=180)
categories = ["Electronics", "Fashion", "Home & Kitchen", "Books"]
cities = ["Delhi", "Mumbai", "Bangalore", "Kolkata", "Chennai"]
payments = ["Credit Card", "UPI", "Debit Card", "Cash on Delivery"]

data = {
    "Order ID": [f"ORD_{1000 + i}" for i in range(n_rows)],
    "Customer ID": [f"CUST_{np.random.randint(100, 180)}" for _ in range(n_rows)],
    "Product Name": np.random.choice(["Laptop", "T-Shirt", "Blender", "Novel", "Headphones"], n_rows),
    "Category": np.random.choice(categories, n_rows),
    "Quantity": np.random.choice([1, 2, 3, 4], n_rows, p=[0.6, 0.25, 0.1, 0.05]),
    "Unit Price": np.random.choice([500, 1200, 2500, 15000], n_rows, p=[0.35, 0.3, 0.2, 0.15]),
    "Order Date": np.random.choice(dates, n_rows),
    "Payment Mode": np.random.choice(payments, n_rows),
    "Delivery Status": np.random.choice(["Delivered", "Cancelled", "Returned", "Pending"], n_rows),
    "City": np.random.choice(cities, n_rows),
}

df = pd.DataFrame(data)
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Total Amount"] = df["Quantity"] * df["Unit Price"]

# 2. Key Performance Indicators (KPIs)
total_revenue = df[df["Delivery Status"] == "Delivered"]["Total Amount"].sum()
total_orders = df["Order ID"].nunique()
avg_order_value = total_revenue / total_orders
total_units_sold = df[df["Delivery Status"] == "Delivered"]["Quantity"].sum()
top_category = df.groupby("Category")["Total Amount"].sum().idxmax()
top_product = df.groupby("Product Name")["Total Amount"].sum().idxmax()

kpi_df = pd.DataFrame({
    "KPI": ["Total Revenue (Delivered)", "Total Orders", "Average Order Value (AOV)", "Total Units Sold", "Top-Selling Category", "Top Product"],
    "Value": [f"₹{total_revenue:,.2f}", str(total_orders), f"₹{avg_order_value:,.2f}", str(total_units_sold), top_category, top_product]
})

print("=== TASK 2: KPI SUMMARY TABLE ===")
print(kpi_df.to_string(index=False))

# 3. Monthly Trend
df["Month"] = df["Order Date"].dt.strftime("%Y-%m")
print("\n=== MONTHLY REVENUE ===")
print(df.groupby("Month")["Total Amount"].sum())
```
